# Loading all the pickle file & model!

```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
## Loading trained model, preprocessing

model = load_model('model.h5')


## Load preprocessing -> encoding & scalar
with open('preprocessor.pkl','rb') as file:
    preprocessor=pickle.load(file)
```

```python
# Example input data
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
## converting input features in Dataframe

input_df = pd.DataFrame([input_data])
```

```python
input_df
```

```python
## Transforming Input data

input_transformed = preprocessor.transform(input_df)
```

```python
input_transformed
```

```python
input_transformed.shape[1]
```

```python
## Predicting churn
model.predict(input_transformed)
```

```python
probability = model.predict(input_transformed)[0][0] ### tensorflow returns predictions as 2D arary that's why [0][0]

print(f"Churn Probability: {probability:.2%}")

if probability > 0.5:
    print("Customer is likely to churn")
else:
    print("Customer is unlikely to churn")
```

```python
## Defining function for all of it!

def predict_churn(input_data):
    input_df = pd.DataFrame([input_data]) ### Converting to DataFrame

    input_transformed = preprocessor.transform(input_df) ### Transforming input

    probability = model.predict(input_transformed)[0][0] ### Predicting

    return probability  ### returning Probability
```

Raw Input
    ↓
DataFrame
    ↓
Preprocessor.transform()
    ↓
ANN Model.predict()
    ↓
Probability
    ↓
Business Decision

```python
z = predict_churn(input_data)
```

```python
## Decission 

if z > 0.5:
    print("Customer is likely to churn")
else:
    print("Customer is unlikely to churn")
```
